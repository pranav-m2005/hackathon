```sql
CREATE TABLE IF NOT EXISTS bronze_customer (
  customer_id     STRING,
  customer_name   STRING,
  contact_number  STRING,
  shipping_city   STRING,
  billing_city    STRING,
  source_file     STRING,
  ingested_at     TIMESTAMP
);

CREATE TABLE IF NOT EXISTS bronze_product (row_id             BIGINT,
  customer_id        STRING,
  merchant_category  STRING,
  amount_aed         STRING,
  items_count        STRING,
  device_type        STRING,
  browser            STRING,
  transaction_id     STRING,
  return_option      STRING,
  source_file        STRING,
  ingested_at        TIMESTAMP
);

CREATE TABLE IF NOT EXISTS bronze_payment (row_id          BIGINT,
  customer_id     STRING,
  payment_method  STRING,
  timestamp_utc   STRING,
  refund_option   STRING,
  source_file     STRING,
  ingested_at     TIMESTAMP
);

create table if not exists silver_customer like bronze_customer;
create table if not exists silver_product like bronze_product;
create table if not exists silver_payment like bronze_payment;
create table if not exists gold_customer like bronze_customer;
create table if not exists gold_product like bronze_product;
create table if not exists gold_payment like bronze_payment;


show tables;
```
